```python
# 1. 回文判断函数
def is_palindrome(s):
    processed = s.replace(" ", "").lower()
    return processed == processed[::-1]


# 2. 字母顺序判断函数
def is_abecedarian(s):
    chars = [c.lower() for c in s if c != ' ']
    if len(chars) <= 1:
        return True
    for i in range(1, len(chars)):
        if chars[i] < chars[i-1]:
            return False
    return True


# 3. 移除元音函数
def remove_vowels(s):
    vowels = {'a', 'e', 'i', 'o', 'u'}
    result = [char for char in s if char.lower() not in vowels]
    return ''.join(result)


# ========== 测试输入与输出 ==========
if __name__ == "__main__":
    # 测试 is_palindrome
    print("===== is_palindrome 测试 =====")
    test1 = ["level", "T A C O cat", "12321", "hello", "Was it a car or a cat I saw"]
    for text in test1:
        output = is_palindrome(text)
        print(f"输入: '{text}'  →  输出: {output}")

    print()

    # 测试 is_abecedarian
    print("===== is_abecedarian 测试 =====")
    test2 = ["adder", "beet", "abcd efgh xyz", "cat", "dog"]
    for text in test2:
        output = is_abecedarian(text)
        print(f"输入: '{text}'  →  输出: {output}")

    print()

    # 测试 remove_vowels
    print("===== remove_vowels 测试 =====")
    test3 = ["cat", "audio", "AEIOUaeiou", "python programming"]
    for text in test3:
        output = remove_vowels(text)
        print(f"输入: '{text}'  →  输出: '{output}'")
```

```python
# 1. 列表反转函数
def list_reverse(lst):
    if not isinstance(lst, list):
        raise TypeError("输入参数必须是列表类型")
    return lst[::-1]


# 2. 判断序列是否升序排列
def is_sorted(seq):
    if len(seq) <= 1:
        return True
    for i in range(1, len(seq)):
        if seq[i] < seq[i - 1]:
            return False
    return True


# ========== 测试输入与输出 ==========
if __name__ == "__main__":
    # 测试 list_reverse
    print("===== list_reverse 测试 =====")
    test_cases1 = [[1, 2, 3], ['a', 'b', 'c', 'd'], [], [True, False, True]]
    for case in test_cases1:
        res = list_reverse(case)
        print(f"输入: {case}  →  输出: {res}")

    # 测试非列表输入（会抛出异常）
    try:
        list_reverse("abc")
    except TypeError as e:
        print(f"输入: 'abc'（非列表） →  抛出错误: {e}")

    print()

    # 测试 is_sorted
    print("===== is_sorted 测试 =====")
    test_cases2 = [
        [1, 2, 3, 4],
        [1, 3, 2, 4],
        [],
        [5],
        "abcde",
        "abdc",
        ('x', 'y', 'z')
    ]
    for case in test_cases2:
        res = is_sorted(case)
        print(f"输入: {case}  →  输出: {res}")
```

```python
# 1. 向量标量乘法
def vec_scalar_mult(t, s):

    if not isinstance(t, tuple):
        raise TypeError("第一个参数必须是元组类型")
    # 校验元组内元素类型
    for elem in t:
        if not isinstance(elem, (int, float)):
            raise TypeError("元组内所有元素必须为整数或浮点数")
    # 校验标量类型
    if not isinstance(s, (int, float)):
        raise TypeError("标量s必须为整数或浮点数")
    # 逐元素相乘并返回新元组
    return tuple(elem * s for elem in t)


# 2. 向量内积计算
def vec_inner_product(t1, t2):
    # 校验两个输入均为元组
    if not isinstance(t1, tuple) or not isinstance(t2, tuple):
        raise TypeError("两个输入参数都必须是元组类型")
    # 校验元素均为数值类型
    for elem in t1:
        if not isinstance(elem, (int, float)):
            raise TypeError("元组内所有元素必须为整数或浮点数")
    for elem in t2:
        if not isinstance(elem, (int, float)):
            raise TypeError("元组内所有元素必须为整数或浮点数")
    # 校验长度一致
    if len(t1) != len(t2):
        raise ValueError("两个向量长度不一致，无法计算内积")
    # 计算内积并以浮点数返回
    return float(sum(x * y for x, y in zip(t1, t2)))


# 3. 矩阵有效性校验
def check_valid_mx(mx):
    # 校验外层是元组
    if not isinstance(mx, tuple):
        return False
    # 校验每一行都是元组
    for row in mx:
        if not isinstance(row, tuple):
            return False
    # 校验所有行长度一致（非空矩阵时）
    if len(mx) > 0:
        col_num = len(mx[0])
        for row in mx:
            if len(row) != col_num:
                return False
    # 校验所有元素为整数或浮点数
    for row in mx:
        for elem in row:
            if not isinstance(elem, (int, float)):
                return False
    return True


# 4. 矩阵与向量乘法
def mx_vec_mult(mx, vec):
    # 复用矩阵校验函数
    if not check_valid_mx(mx):
        raise TypeError("输入的矩阵不是有效的元组矩阵")
    # 校验向量类型与元素
    if not isinstance(vec, tuple):
        raise TypeError("向量参数必须是元组类型")
    for elem in vec:
        if not isinstance(elem, (int, float)):
            raise TypeError("向量内所有元素必须为整数或浮点数")
    # 校验维度匹配：矩阵列数 = 向量长度
    if len(mx) > 0 and len(mx[0]) != len(vec):
        raise ValueError("矩阵列数与向量长度不匹配，无法完成乘法")
    # 逐行计算内积，返回结果元组
    result = []
    for row in mx:
        result.append(sum(x * y for x, y in zip(row, vec)))
    return tuple(result)


# ========== 测试输入与输出 ==========
if __name__ == "__main__":
    # 测试 vec_scalar_mult
    print("===== vec_scalar_mult 测试 =====")
    print(f"输入: t=(1, 2, 3), s=2  →  输出: {vec_scalar_mult((1, 2, 3), 2)}")
    print(f"输入: t=(0.5, 1.5), s=4  →  输出: {vec_scalar_mult((0.5, 1.5), 4)}")
    try:
        vec_scalar_mult([1, 2, 3], 2)
    except TypeError as e:
        print(f"输入: t=[1,2,3]（非元组） →  抛出错误: {e}")

    print()

    # 测试 vec_inner_product
    print("===== vec_inner_product 测试 =====")
    print(f"输入: (1, 2, 3) · (4, 5, 6)  →  输出: {vec_inner_product((1, 2, 3), (4, 5, 6))}")
    print(f"输入: (2.0, 3.0) · (1.0, 4.0)  →  输出: {vec_inner_product((2.0, 3.0), (1.0, 4.0))}")
    try:
        vec_inner_product((1, 2), (3, 4, 5))
    except ValueError as e:
        print(f"输入: 长度不同的两个元组  →  抛出错误: {e}")

    print()

    # 测试 check_valid_mx
    print("===== check_valid_mx 测试 =====")
    print(f"输入: ((1,2), (3,4), (5,6))  →  输出: {check_valid_mx(((1,2), (3,4), (5,6)))}")
    print(f"输入: [(1,2), (3,4)]（外层是列表）  →  输出: {check_valid_mx([(1,2), (3,4)])}")
    print(f"输入: ((1,2), (3,4,5))（行长度不一致）  →  输出: {check_valid_mx(((1,2), (3,4,5)))}")
    print(f"输入: ((1, 'a'), (2, 3))（含非数值元素）  →  输出: {check_valid_mx(((1, 'a'), (2, 3)))}")

    print()

    # 测试 mx_vec_mult
    print("===== mx_vec_mult 测试 =====")
    mx = ((1, 2), (3, 4), (5, 6))
    vec = (7, 8)
    print(f"矩阵{mx} × 向量{vec}  →  输出: {mx_vec_mult(mx, vec)}")
    try:
        mx_vec_mult(mx, (1, 2, 3))
    except ValueError as e:
        print(f"输入: 维度不匹配的矩阵与向量  →  抛出错误: {e}")
```

```python
class Vector:
    def __init__(self, dim, elements=None):
        # 1. 校验维度类型与合法性
        if not isinstance(dim, int):
            raise TypeError("维度必须为整数类型")
        if dim < 0:
            raise ValueError("维度不能为负数")

        # 未提供元素时，默认初始化为全零向量
        if elements is None:
            elements = [0] * dim

        # 2. 校验元素容器类型
        if not isinstance(elements, (list, tuple)):
            raise TypeError("向量元素必须以列表或元组形式传入")

        # 3. 校验元素个数与维度一致
        if len(elements) != dim:
            raise ValueError("元素个数与向量维度不匹配")

        # 4. 校验所有元素为数值类型
        for e in elements:
            if not isinstance(e, (int, float)):
                raise TypeError("向量元素必须为整数或浮点数")

        # 实例属性：维度、元素（转为元组保证不可变）
        self.dim = dim
        self.elems = tuple(elements)

    # ========== 比较运算符（） ==========
    def __eq__(self, other):
        if not isinstance(other, Vector):
            return False
        if self.dim != other.dim:
            raise ValueError("维度不同的向量无法进行相等比较")
        return self.elems == other.elems

    def __lt__(self, other):
        if not isinstance(other, Vector):
            raise TypeError("比较对象必须是Vector类型")
        if self.dim != other.dim:
            raise ValueError("维度不同的向量无法进行大小比较")
        # 按元组字典序比较
        return self.elems < other.elems

    def __le__(self, other):
        return self < other or self == other

    def __ge__(self, other):
        return other <= self

    # ========== 内积方法（） ==========
    def dot(self, other):
        if not isinstance(other, Vector):
            raise TypeError("内积计算的参数必须是Vector类型")
        if self.dim != other.dim:
            raise ValueError("维度不同的向量无法计算内积")
        return sum(a * b for a, b in zip(self.elems, other.elems))

    # ========== 乘法运算符（） ==========
    def __mul__(self, other):
        # 标量右乘：向量 * 标量
        if isinstance(other, (int, float)):
            new_elems = [e * other for e in self.elems]
            return Vector(self.dim, new_elems)
        # 向量逐元素相乘
        elif isinstance(other, Vector):
            if self.dim != other.dim:
                raise ValueError("维度不同的向量无法逐元素相乘")
            new_elems = [a * b for a, b in zip(self.elems, other.elems)]
            return Vector(self.dim, new_elems)
        else:
            raise TypeError("乘法参数必须是数值标量或Vector对象")

    def __rmul__(self, other):
        # 标量左乘：标量 * 向量，利用交换律复用右乘逻辑
        return self * other

    # 辅助打印，方便查看结果
    def __repr__(self):
        return f"Vector({self.dim}, {list(self.elems)})"


# ========== 测试用例与输出 ==========
if __name__ == "__main__":
    print("===== 1. 初始化测试 =====")
    v1 = Vector(2, (2, 4))
    v2 = Vector(2, (2, 5))
    v3 = Vector(3, [1, 2, 3])
    v_zero = Vector(2)  # 默认全零
    print(f"v1 = {v1}")
    print(f"v2 = {v2}")
    print(f"v3 = {v3}")
    print(f"默认零向量 v_zero = {v_zero}")

    print("\n===== 2. 比较运算测试 =====")
    print(f"v1 == v2 : {v1 == v2}")
    print(f"v1 < v2  : {v1 < v2}")
    print(f"v1 <= v2 : {v1 <= v2}")
    print(f"v2 >= v1 : {v2 >= v1}")

    print("\n===== 3. 内积测试 =====")
    print(f"v1 · v2 = {v1.dot(v2)}")
    print(f"v3 · Vector(3, (4,5,6)) = {v3.dot(Vector(3, (4,5,6)))}")

    print("\n===== 4. 乘法测试 =====")
    print(f"v1 * 2 = {v1 * 2}")
    print(f"3 * v1 = {3 * v1}")
    print(f"v1 * v2 = {v1 * v2}")
    print(f"Vector(2, (2,3)) * Vector(2, (4,5)) = {Vector(2, (2,3)) * Vector(2, (4,5))}")

    print("\n===== 5. 异常测试示例 =====")
    try:
        Vector(-2, [1,2])
    except ValueError as e:
        print(f"负维度异常: {e}")
    try:
        v1 < v3
    except ValueError as e:
        print(f"维度不同比较异常: {e}")
```
