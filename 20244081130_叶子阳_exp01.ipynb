### 实验2：函数、条件、循环与递归

本实验的核心是「把逻辑封装成函数」：用条件分支做分类，用两种循环验证同一结论，
用递归与迭代对照同一算法的两种写法，最后认识 `*args` 这样的可变位置参数。

````python
# 实验2-1：成绩等级划分，越界抛 ValueError
def score_level(score):
    """把 0~100 的分数转换为等级；超出范围抛出 ValueError。"""
    if not 0 <= score <= 100:
        raise ValueError(f"分数必须在 0~100 之间，收到: {score!r}")
    if score >= 90:
        return "优秀"
    elif score >= 80:
        return "良好"
    elif score >= 60:
        return "及格"
    else:
        return "不及格"


print("边界值测试：")
for sc in [100, 95, 90, 89.9, 85, 80, 79.9, 60, 59.9, 0]:
    print(f"  score_level({sc:>5}) = {score_level(sc)}")

print()
print("越界测试：")
for bad in [-1, 101]:
    try:
        score_level(bad)
    except ValueError as exc:
        print(f"  score_level({bad}) -> 捕获 {type(exc).__name__}: {exc}")
````

````python
# 实验2-2：for 与 while 求 1~100 的整数和
total_for = 0
for i in range(1, 101):
    total_for += i

total_while = 0
j = 1
while j <= 100:
    total_while += j
    j += 1

print("for   循环求和 =", total_for)
print("while 循环求和 =", total_while)
print("两种实现结果一致:", total_for == total_while)
print("用公式 n(n+1)/2 验证:", 100 * 101 // 2)
````

````python
# 实验2-3：递归版与迭代版阶乘，约定 0! = 1
def factorial_recursive(n):
    """递归版阶乘；n 必须是非负整数。"""
    if not isinstance(n, int) or n < 0:
        raise ValueError(f"n 必须是非负整数，收到: {n!r}")
    if n == 0:
        return 1
    return n * factorial_recursive(n - 1)


def factorial_iterative(n):
    """迭代版阶乘；n 必须是非负整数。"""
    if not isinstance(n, int) or n < 0:
        raise ValueError(f"n 必须是非负整数，收到: {n!r}")
    result = 1
    for k in range(2, n + 1):
        result *= k
    return result


import math

print(f"{'n':>3} {'递归':>12} {'迭代':>12} {'math.factorial':>16}  三者一致")
for n in [0, 1, 5, 10]:
    r = factorial_recursive(n)
    it = factorial_iterative(n)
    m = math.factorial(n)
    print(f"{n:>3} {r:>12} {it:>12} {m:>16}  {r == it == m}")

print()
try:
    factorial_recursive(-1)
except ValueError as exc:
    print("传入 -1 -> 捕获", type(exc).__name__, ":", exc)
````

````python
# 实验2-4：summary() 返回元组，调用处用元组解包接收
def summary(values):
    """返回 (最小值, 最大值, 平均值) 三元组；空列表抛出 ValueError。"""
    if not values:
        raise ValueError("values 不能为空")
    return (min(values), max(values), sum(values) / len(values))


data = [85, 58, 92, 76, 60, 89]
low, high, avg = summary(data)          # 元组解包
print("原始数据:", data)
print(f"解包结果: 最小值 = {low}, 最大值 = {high}, 平均值 = {avg:.2f}")
print("不解包的完整元组:", summary(data))

print()
try:
    summary([])
except ValueError as exc:
    print("传入空列表 -> 捕获", type(exc).__name__, ":", exc)
````

````python
# 实验2-5：sum_all(*args) 处理任意个位置参数
def sum_all(*args):
    """返回所有参数之和；无参数时返回 0。"""
    print(f"    函数内收到的 args = {args}  (类型: {type(args).__name__})")
    return sum(args)


print("无参数   sum_all()           =", sum_all())
print("单个参数 sum_all(10)         =", sum_all(10))
print("多个参数 sum_all(1, 2, 3, 4) =", sum_all(1, 2, 3, 4))
````

````python
# 实验3-1：用 zip() 构建字典，再用 items() 逐项输出
names = ["Alice", "Bob", "Carol", "David", "Eva", "Frank"]
scores = [85, 58, 92, 76, 60, 89]

score_dict = dict(zip(names, scores))
print("zip() 构建的字典:", score_dict)
print()
print("items() 逐项输出:")
for name, score in score_dict.items():
    print(f"  {name:<6} -> {score}")
````

````python
# 实验3-5：alias 与 copy() 的区别（使用单独的演示列表，不动上面的 scores）
original = [10, 20, 30]
alias = original            # 直接赋值：只是给同一个对象再起一个名字
copied = original.copy()    # copy()：新建一个列表对象

alias[0] = 999              # 通过 alias 修改元素

print("执行 alias[0] = 999 之后：")
print("  original =", original)
print("  alias    =", alias, "   与 original 是同一对象:", original is alias)
print("  copied   =", copied, "   与 original 是同一对象:", original is copied)
print()
print("结论:")
print("  alias = original 只是增加了一个引用，两者指向同一个列表对象，所以一起变；")
print("  copy() 产生的副本是独立对象，修改副本或修改原列表都不会互相影响。")
````

````python
# 实验3-6：元组的不可变性，以及 zip() 遇到不等长序列的表现
student = ("Alice", 85, "本24数信01班")
print("学生记录元组:", student, " 长度:", len(student))

try:
    student[1] = 90
except TypeError as exc:
    print("尝试 student[1] = 90 失败")
    print("  异常类型:", type(exc).__name__)
    print("  异常信息:", exc)
    print("  原因: 元组是不可变序列，不能对元素重新赋值。")

rebuilt = student[:1] + (90,) + student[2:]   # 需要“修改”时只能重建元组
print("重建后的元组:", rebuilt)
print()

a = ["Alice", "Bob", "Carol"]
b = [85, 58]
print("长度不同的两个序列:", a, "和", b)
print("zip(a, b) 的结果   :", list(zip(a, b)))
print("结论: zip() 以较短的序列为准，多余元素被丢弃，不报错。")
````

### 实验4：文件读写、异常处理与模块

本实验模拟一次真实的数据清洗流程：把原始文本落盘 → 逐行解析并校验 → 把无效记录挡在外面
并记录原因 → 输出统计报告 → 用 pickle 保存结构化结果 → 把解析函数抽成可复用模块。

数据文件 `exp01_data/students.txt` 中故意混入了 3 条无效记录（`Grace,abc` 成绩非整数、
`Henry,105` 成绩越界、`Incomplete` 字段数不足），用来验证异常处理逻辑。

````python
# 实验4-1：查看工作目录，用 os.path.join() 构造相对路径并建目录
import os

print("当前工作目录 os.getcwd():", os.getcwd())

data_dir = os.path.join(".", "exp01_data")     # 相对路径，位于 Notebook 同级目录
os.makedirs(data_dir, exist_ok=True)           # 已存在也不报错
print("数据目录相对路径:", data_dir)
print("数据目录绝对路径:", os.path.abspath(data_dir))
print("目录是否已存在  :", os.path.isdir(data_dir))
````

````python
# 实验4-2：写入 students.txt（UTF-8），再读取并按字符串方法解析
students_txt = os.path.join(data_dir, "students.txt")

raw_lines = [
    "name,score",
    "Alice,85",
    "Bob,58",
    "Carol,92",
    "David,76",
    "Eva,60",
    "Frank,89",
    "Grace,abc",
    "Henry,105",
    "Incomplete",
]
raw_text = "\n".join(raw_lines) + "\n"


def read_text(path):
    """按 UTF-8 读取整个文本文件。"""
    with open(path, "r", encoding="utf-8") as fh:
        return fh.read()


with open(students_txt, "w", encoding="utf-8") as fh:
    fh.write(raw_text)

print("已写入:", os.path.abspath(students_txt), " 字节数:", os.path.getsize(students_txt))
print("-" * 52)

with open(students_txt, "r", encoding="utf-8") as fh:
    for lineno, line in enumerate(fh, start=1):
        line = line.rstrip("\n")
        if lineno == 1:
            print(f"[{lineno}] 跳过表头: {line}")
            continue
        if not line.strip():
            continue
        fields = line.split(",")
        print(f"[{lineno}] 字符串解析 -> 字段数={len(fields)}, 字段={fields}")
````

````python
# 实验4-3：parse_student() —— 把一行解析成字典，无效记录抛 ValueError
def parse_student(line):
    """把 'name,score' 一行解析为字典；无效记录抛出 ValueError。

    校验规则：字段数量必须为 2、姓名不能为空、成绩能转成整数且落在 0~100。
    """
    fields = [f.strip() for f in line.strip().split(",")]
    if len(fields) != 2:
        raise ValueError(f"字段数量应为 2，实际为 {len(fields)}")
    name, score_text = fields
    if not name:
        raise ValueError("姓名为空")
    try:
        score = int(score_text)
    except ValueError as exc:
        raise ValueError(f"成绩不是整数: {score_text!r}") from exc
    if not 0 <= score <= 100:
        raise ValueError(f"成绩超出 0~100 范围: {score}")
    return {"name": name, "score": score}


print("合法记录测试：")
for line in ["Alice,85", "  Bob , 58 ", "Eva,60"]:
    print(f"  {line!r:<16} -> {parse_student(line)}")

print()
print("非法记录测试：")
for line in ["Grace,abc", "Henry,105", "Incomplete", ",77"]:
    try:
        parse_student(line)
    except ValueError as exc:
        print(f"  {line!r:<16} -> ValueError: {exc}")
````

````python
# 实验4-4：逐行处理，捕获无效记录并记录行号与原因
valid_records = []
invalid_records = []

with open(students_txt, "r", encoding="utf-8") as fh:
    for lineno, line in enumerate(fh, start=1):
        if lineno == 1:                 # 跳过表头
            continue
        if not line.strip():            # 跳过空行
            continue
        try:
            valid_records.append(parse_student(line))
        except ValueError as exc:
            invalid_records.append({"line": lineno, "text": line.strip(), "reason": str(exc)})

print("有效记录数:", len(valid_records))
for rec in valid_records:
    print("   ", rec)

print()
print("无效记录数:", len(invalid_records))
for bad in invalid_records:
    print(f"    第 {bad['line']:>2} 行  {bad['text']!r:<16} 原因: {bad['reason']}")
````

````python
# 实验4-5：把统计结果写入 report.txt，并演示 w 与 a 两种模式
report_txt = os.path.join(data_dir, "report.txt")

n_valid = len(valid_records)
n_invalid = len(invalid_records)
valid_scores = [r["score"] for r in valid_records]
avg_score = sum(valid_scores) / len(valid_scores)
pass_rate = sum(1 for s in valid_scores if s >= 60) / len(valid_scores)

content = (
    "实验01 数据处理报告\n"
    "====================\n"
    f"有效记录数: {n_valid}\n"
    f"无效记录数: {n_invalid}\n"
    f"平均分    : {avg_score:.2f}\n"
    f"及格率    : {pass_rate:.2%}\n"
)

# --- w 模式：写入前会清空文件原有全部内容 ---
with open(report_txt, "w", encoding="utf-8") as fh:
    fh.write(content)
print("=== ① 用 w 模式写入后 ===")
print(read_text(report_txt))

# --- a 模式：在文件末尾追加，原有内容完整保留 ---
with open(report_txt, "a", encoding="utf-8") as fh:
    fh.write("[a 模式演示] 这一行是追加写入的\n")
print("=== ② 用 a 模式追加后（原有内容仍在，新内容在末尾）===")
print(read_text(report_txt))

# --- 再用 w 模式重写：刚才追加的那行会被清空 ---
with open(report_txt, "w", encoding="utf-8") as fh:
    fh.write(content)
print("=== ③ 再次用 w 模式重写后（追加行已消失）===")
print(read_text(report_txt))

print("结论:")
print("  w 模式会先截断（清空）文件再写入，原有内容全部丢失；")
print("  a 模式只把新内容追加到末尾，原有内容不受影响。")
````

````python
# 实验4-6：用 pickle 保存有效记录，再读回验证一致性
import pickle

pkl_path = os.path.join(data_dir, "students.pkl")
with open(pkl_path, "wb") as fh:
    pickle.dump(valid_records, fh)

with open(pkl_path, "rb") as fh:
    restored_records = pickle.load(fh)

print("原数据        :", valid_records)
print("pickle 恢复后 :", restored_records)
print("两者相等      :", restored_records == valid_records)
print("文件大小      :", os.path.getsize(pkl_path), "字节")
````

### 实验5：类、继承与对象复制

一点：**直接赋值 / 浅拷贝 / 深拷贝** 在对象图中到底复制了什么。

````python
# 实验5-1：Student 类
class Student:
    """学生类：保存姓名与成绩，并判断是否及格。"""

    def __init__(self, name, score):
        self.name = name
        self.score = score

    def is_passed(self):
        """成绩不低于 60 分视为及格。"""
        return self.score >= 60

    def __str__(self):
        status = "及格" if self.is_passed() else "不及格"
        return f"Student(name={self.name!r}, score={self.score}, {status})"


demo = Student("Alice", 85)
print("print(obj) 调用 __str__  :", demo)
print("is_passed()              :", demo.is_passed())
print("repr(obj)                :", repr(demo))
````

````python
# 实验5-3：GraduateStudent 继承 Student，增加 research_topic 并重写 __str__
class GraduateStudent(Student):
    """研究生：在 Student 基础上增加研究课题。"""

    def __init__(self, name, score, research_topic):
        super().__init__(name, score)          # 用 super() 初始化父类属性
        self.research_topic = research_topic

    def __str__(self):                          # 重写父类 __str__
        status = "及格" if self.is_passed() else "不及格"
        return (f"GraduateStudent(name={self.name!r}, score={self.score}, "
                f"topic={self.research_topic!r}, {status})")


gs = GraduateStudent("Eva", 60, "机器学习")
print("print(obj)        :", gs)
print("继承来的 is_passed:", gs.is_passed())
print("issubclass 关系   :", issubclass(GraduateStudent, Student))
print("isinstance 关系   :", isinstance(gs, Student))
````

````python
# 实验5-4：直接赋值 / copy.copy() / copy.deepcopy() 的对比
# 每次都从新建的原始对象开始，保证三次比较互不干扰
import copy


def make_origin():
    """返回一个包含嵌套列表的新对象。"""
    return {"name": "Alice", "scores": [85, 90]}


# ① 直接赋值：只是同一个对象的别名
o1 = make_origin()
assigned = o1
assigned["scores"].append(100)
print("① 直接赋值  assigned = o1")
print("   修改 assigned['scores'] 后 -> original =", o1, " assigned =", assigned)
print("   同一对象:", o1 is assigned)
print()

# ② copy.copy()：浅拷贝，外层是新对象，内层可变对象仍被共享
o2 = make_origin()
shallow = copy.copy(o2)
shallow["scores"].append(100)
print("② 浅拷贝    shallow = copy.copy(original)")
print("   修改 shallow['scores'] 后 -> original =", o2, " shallow =", shallow)
print("   外层是同一对象:", o2 is shallow, "  内层列表是同一对象:", o2["scores"] is shallow["scores"])
print()

# ③ copy.deepcopy()：深拷贝，内外层都是新对象
o3 = make_origin()
deep = copy.deepcopy(o3)
deep["scores"].append(100)
print("③ 深拷贝    deep = copy.deepcopy(original)")
print("   修改 deep['scores'] 后 -> original =", o3, " deep =", deep)
print("   外层是同一对象:", o3 is deep, "  内层列表是同一对象:", o3["scores"] is deep["scores"])
print()
print("结论:")
print("  直接赋值      —— 两个名字指向同一个对象，改动互相可见；")
print("  copy.copy()   —— 只复制最外层，嵌套的可变对象仍然共享，改内层会波及原对象；")
print("  copy.deepcopy()—— 递归复制整张对象图，与原对象完全独立。")
````

### 实验小结

1. **基本运算与类型**：`/` 恒得浮点，`//` 向下取整，`%` 取余，`**` 求幂；`int()` 转换浮点数时直接截断而非四舍五入，非法字符串会抛 `ValueError`。
2. **字符串不可变**：`s[0] = 'J'` 会抛 `TypeError`，所有“修改”操作本质上都是构造新字符串。
3. **函数与流程控制**：`for` 与 `while` 得到相同结果；递归写法更贴近数学定义，但受递归深度限制；`*args` 让函数接收任意个位置参数。
4. **容器与统计**：`zip()` 以较短序列为准；`=` 是起别名而 `copy()` 才产生独立副本；元组不可变，需要“修改”时只能重建。
5. **文件与异常**：`with open(..., encoding="utf-8")` 保证正确关闭与中文编码；`w` 模式清空重写、`a` 模式末尾追加；异常按记录粒度捕获后，坏数据不会中断整批处理。
6. **模块化**：把 `parse_student()` 抽到 `student_utils.py` 后可通过 `import` 复用，说明代码可以跨 Notebook 共享。
7. **类与对象**：继承通过 `super().__init__()` 复用父类初始化逻辑，重写 `__str__()` 改变对象的字符串表示；浅拷贝与深拷贝的差别只有在对象存在嵌套可变成员时才会暴露。
