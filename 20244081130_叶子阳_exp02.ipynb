```python
import numpy as np

# 设置随机种子，保证结果可复现
np.random.seed(42)

# 任务1：生成模拟成绩并计算统计量
print("==================== 任务1 ====================")
# 生成1000个均值70、标准差10的正态分布成绩
scores = np.random.normal(loc=70, scale=10, size=1000)
# 将数值限制在0~100范围内
scores = np.clip(scores, a_min=0, a_max=100)

# 计算各项统计指标
sample_count = scores.size
min_val = scores.min()
max_val = scores.max()
mean_val = scores.mean()
std_val = scores.std()
range_val = max_val - min_val  # 极差

print(f"样本数: {sample_count}")
print(f"最小值: {min_val:.2f}")
print(f"最大值: {max_val:.2f}")
print(f"均值: {mean_val:.2f}")
print(f"标准差: {std_val:.2f}")
print(f"极差: {range_val:.2f}")


# 任务2：均匀抽样与加权抽样对比
print("\n==================== 任务2 ====================")
population = np.arange(1, 11)  # 1到10的整数
print(f"总体: {population}")

# 1. 均匀抽样（不放回）：每个元素被抽取的概率相等
sample_uniform = np.random.choice(population, size=5, replace=False)
print(f"均匀抽样结果: {sample_uniform}")

# 2. 加权抽样（不放回）：按整数大小成比例加权，数值越大被抽中概率越高
weights = population / population.sum()  # 权重与数值大小成正比
sample_weighted = np.random.choice(population, size=5, replace=False, p=weights)
print(f"按数值大小加权抽样结果: {sample_weighted}")
print("说明：均匀抽样每个元素概率均为1/10；加权抽样中数值越大，被抽中的概率越高")


# 任务3：随机排列（保留原数组）与原地打乱对比
print("\n==================== 任务3 ====================")
arr_original = np.arange(10)
print(f"操作前原数组: {arr_original}")

# 方式1：np.random.permutation —— 返回新数组，不修改原数组
arr_permuted = np.random.permutation(arr_original)
print(f"\npermutation 后原数组: {arr_original}")
print(f"permutation 返回的新数组: {arr_permuted}")

# 方式2：np.random.shuffle —— 原地修改数组，无返回值
arr_for_shuffle = arr_original.copy()  # 复制一份避免修改原数组
np.random.shuffle(arr_for_shuffle)
print(f"\nshuffle 后原数组（副本）: {arr_for_shuffle}")
print("区别：permutation生成新数组，原数组保持不变；shuffle直接修改原数组，属于原地操作")


# 任务4：缺失值处理与统计对比
print("\n==================== 任务4 ====================")
# 复制成绩数组，将第10、20、30个值设为nan（Python为0基索引，对应下标9、19、29）
scores_nan = scores.copy()
scores_nan[9] = np.nan
scores_nan[19] = np.nan
scores_nan[29] = np.nan

# 统计缺失值个数
nan_count = np.isnan(scores_nan).sum()
print(f"缺失值个数: {nan_count}")

# 普通均值（含nan）
mean_normal = np.mean(scores_nan)
# 忽略缺失值的均值
mean_nan = np.nanmean(scores_nan)

print(f"普通均值（含nan）: {mean_normal}")
print(f"忽略缺失值后的均值: {mean_nan:.2f}")

print("\nnan对普通统计函数的影响：")
print("普通统计函数（np.mean、np.std、np.sum等）遇到nan时，运算结果会直接变为nan，因为NaN与任何数值运算结果都是NaN。")
print("需要使用np.nanmean、np.nanstd、np.nansum等带nan前缀的函数，才能自动忽略缺失值进行统计。")


# 任务5：整理为内存中的表格变量
print("\n==================== 任务5 ====================")
# 生成序号（从1开始）
ids = np.arange(1, len(scores_nan) + 1)

# 构造结构化数组（内存中的表格，包含「序号」和「成绩」两列）
score_table = np.array(
    list(zip(ids, scores_nan)),
    dtype=[("序号", int), ("成绩", float)]
)

# 打印前5行预览
print("内存表格变量（前5行预览）:")
for row in score_table[:5]:
    print(f"序号: {row['序号']}, 成绩: {row['成绩']:.2f}")

print("\n说明：score_table为内存中的结构化表格变量，包含序号和带缺失值的成绩，可直接供后续实验使用，无需写入磁盘。")
```

```python
import numpy as np
from scipy.linalg import solve
from scipy.optimize import root, minimize
from scipy.stats import ttest_ind
import matplotlib.pyplot as plt

# 设置Matplotlib中文显示，避免乱码
plt.rcParams['font.sans-serif'] = ['SimHei']
plt.rcParams['axes.unicode_minus'] = False

# ==================================================
# 任务1：求解线性方程组
# ==================================================
print("==================== 任务1：线性方程组求解 ====================")
# 方程组：3x + y = 9；x + 2y = 8
A = np.array([[3, 1],
              [1, 2]])
b = np.array([9, 8])

# 求解
x = solve(A, b)
print(f"方程组的解：x = {x[0]:.2f}, y = {x[1]:.2f}")

# NumPy代入验证
verify = np.dot(A, x)
print(f"代入验证 Ax 结果：{verify}")
print(f"与右侧 b 的差值：{np.round(verify - b, 6)}")


# ==================================================
# 任务2：求方程 x² - 2 = 0 的根
# ==================================================
print("\n==================== 任务2：方程求根 ====================")
# 定义目标函数
def func(x):
    return x ** 2 - 2

# 求正根，初始猜测值为2
sol_pos = root(func, x0=2)
# 求负根，初始猜测值为-2
sol_neg = root(func, x0=-2)

print(f"正根求解是否成功：{sol_pos.success}")
print(f"正根：{sol_pos.x[0]:.6f}，对应函数值：{sol_pos.fun[0]:.6e}")

print(f"\n负根求解是否成功：{sol_neg.success}")
print(f"负根：{sol_neg.x[0]:.6f}，对应函数值：{sol_neg.fun[0]:.6e}")
print("解析解为 ±√2 ≈ ±1.414214，数值求解结果与解析解一致，函数值均接近0")


# ==================================================
# 任务3：求函数最小值
# ==================================================
print("\n==================== 任务3：函数最小值求解 ====================")
# 目标函数 f(x) = x² + x + 2
def f_min(x):
    return x ** 2 + x + 2

# 数值求解最小值，初始猜测为0
res = minimize(f_min, x0=0)

print(f"数值求解极小点：x = {res.x[0]:.6f}")
print(f"数值求解最小函数值：f(x) = {res.fun:.6f}")

# 解析解：二次函数极小点 x = -b/(2a) = -1/2 = -0.5
x_analytic = -0.5
f_analytic = x_analytic ** 2 + x_analytic + 2
print(f"\n解析解极小点：x = {x_analytic}")
print(f"解析解最小函数值：f(x) = {f_analytic}")
print("数值求解结果与解析解高度吻合")


# ==================================================
# 任务4：独立样本t检验
# ==================================================
print("\n==================== 任务4：独立样本t检验 ====================")
# 固定随机种子，保证结果可复现
np.random.seed(42)

# 生成两组独立样本
sample1 = np.random.normal(loc=0, scale=1, size=100)  # 均值0，标准差1
sample2 = np.random.normal(loc=1, scale=1, size=100)  # 均值1，标准差1

# 独立样本t检验
t_stat, p_value = ttest_ind(sample1, sample2)

print(f"t统计量：{t_stat:.4f}")
print(f"p值：{p_value:.4e}")

# 显著性判断
alpha = 0.05
if p_value < alpha:
    conclusion = "拒绝原假设，两组样本均值存在显著差异"
else:
    conclusion = "不拒绝原假设，两组样本均值无显著差异"
print(f"在0.05显著性水平下：{conclusion}")


# ==================================================
# 任务5：Matplotlib可视化
# ==================================================
print("\n==================== 任务5：数据可视化 ====================")
# 先生成实验2的带缺失值成绩，用于绘图
np.random.seed(42)
scores = np.random.normal(70, 10, 1000)
scores = np.clip(scores, 0, 100)
scores_nan = scores.copy()
scores_nan[[9, 19, 29]] = np.nan
valid_scores = scores_nan[~np.isnan(scores_nan)]  # 提取有效成绩

# 1. 成绩直方图
plt.figure(figsize=(8, 5))
plt.hist(valid_scores, bins=20, edgecolor='black', alpha=0.7)
plt.title('模拟成绩分布直方图')
plt.xlabel('成绩')
plt.ylabel('频数')
plt.grid(axis='y', alpha=0.3)

# 2. 成绩箱线图
plt.figure(figsize=(6, 5))
plt.boxplot(valid_scores, labels=['成绩'])
plt.title('模拟成绩箱线图')
plt.ylabel('成绩')
plt.grid(axis='y', alpha=0.3)

# 3. 函数折线图（同一坐标系）
x = np.linspace(0, 5, 100)
y1 = x
y2 = x ** 2
y3 = x ** 3

plt.figure(figsize=(8, 5))
plt.plot(x, y1, label='y = x', linewidth=2)
plt.plot(x, y2, label='y = x²', linewidth=2)
plt.plot(x, y3, label='y = x³', linewidth=2)
plt.title('幂函数对比折线图 (0 ≤ x ≤ 5)')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.grid(alpha=0.3)

# 显示所有图像（Notebook中直接展示，不保存文件）
plt.show()
print("已生成直方图、箱线图和函数折线图，在Notebook中直接显示")
```
