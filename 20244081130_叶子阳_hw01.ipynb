```python
def gcd(m, n):
    while n != 0:
        m, n = n, m % n
    return m

print(gcd(2018, 2019))
print(gcd(1200, 300))
print(gcd(5040, 60))
```

```python
import math

# 1. 极限形式近似自然常数e
def euler_limit(n):
    """通过伯努利极限公式 (1 + 1/n)^n 近似e，返回浮点结果"""
    return (1 + 1 / n) ** n


# 2. 无穷级数前n项和近似e
def euler_infinite_sum(n):
    """计算e的泰勒展开前n项之和（1-索引），n=0返回0，n=1返回1"""
    if n == 0:
        return 0.0
    total = 0.0
    factorial = 1  # 初始为 0! = 1
    for k in range(n):
        total += 1 / factorial
        factorial *= (k + 1)  # 递推计算下一个阶乘，避免重复计算
    return total


# 3. 按精度epsilon逼近e
def euler_approx(epsilon):
    """通过无穷级数累加，直到近似值与真实e的误差小于epsilon"""
    real_e = math.exp(1)
    total = 0.0
    factorial = 1
    k = 0
    # 误差不满足要求时，继续累加新的项
    while abs(total - real_e) >= epsilon:
        total += 1 / factorial
        factorial *= (k + 1)
        k += 1
    return total


# 4. 打印前n项级数和的对照表
def print_euler_sum_table(n):
    """逐行打印 k=1 到 k=n 对应的e近似值"""
    for k in range(1, n + 1):
        print(euler_infinite_sum(k))


# ------------------- 键盘输入与结果输出 -------------------
# 1. 调用 euler_limit
n_limit = int(input("请输入 euler_limit 的参数 n（整数）："))
res1 = euler_limit(n_limit)
print(f"euler_limit({n_limit}) = {res1}\n")

# 2. 调用 euler_infinite_sum
n_sum = int(input("请输入 euler_infinite_sum 的参数 n（非负整数）："))
res2 = euler_infinite_sum(n_sum)
print(f"euler_infinite_sum({n_sum}) = {res2}\n")

# 3. 调用 euler_approx
eps = float(input("请输入 euler_approx 的精度 epsilon（浮点数）："))
res3 = euler_approx(eps)
print(f"精度为 {eps} 时，e 的近似值 = {res3}")
print(f"与真实值 e 的绝对误差：{abs(res3 - math.exp(1))}\n")

# 4. 调用 print_euler_sum_table
n_table = int(input("请输入 print_euler_sum_table 的参数 n（正整数）："))
print(f"euler_infinite_sum 从 1 到 {n_table} 的结果表：")
print_euler_sum_table(n_table)
```

```python
def is_power_of_2(n):
    if n == 1:
        return True
    while n > 1:
        if n % 2 != 0:
            return False
        n = n // 2
    return True


def is_power(b, n):
    if n == 1:
        return True
    if b == 1:
        return False
    while n > 1:
        if n % b != 0:
            return False
        n = n // b
    return True


# ------------------- 输入输出测试 -------------------
# 测试 is_power_of_2
num1 = int(input("请输入一个正整数，判断是否为2的幂："))
print(f"{num1} 是2的幂吗？{is_power_of_2(num1)}")

# 测试 is_power
b_val = int(input("\n请输入底数b（正整数）："))
n_val = int(input("请输入待判断的数n（正整数）："))
print(f"{n_val} 是 {b_val} 的幂吗？{is_power(b_val, n_val)}")
```
