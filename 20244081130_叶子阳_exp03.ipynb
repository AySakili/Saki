```python
import re

# ========== 任务1 ==========
text1 = "cat on mat; later: cats and dogs"
print("===== 任务1 =====")
res_d = re.findall(r"\d+", text1)
print(r"\d+ 匹配数字：", res_d)
res_s = re.findall(r"\s+", text1)
print(r"\s+ 匹配空白：", res_s)
res_bcat = re.findall(r"\bcat\b", text1)
print(r"\bcat\b 匹配完整单词cat：", res_bcat)

# ========== 任务2 ==========
print("\n===== 任务2 =====")
text2 = "cat dog bird CAT"
res_catdog = re.findall(r"cat|dog", text2)
print(r"cat|dog 匹配：", res_catdog)

r1 = re.findall(r"a|aa|aaa", "aaaa")
print(r're.findall(r"a|aa|aaa", "aaaa") 结果：', r1)
r2 = re.findall(r"aaa|aa|a", "aaaa")
print(r're.findall(r"aaa|aa|a", "aaaa") 结果：', r2)

# ========== 任务3 ==========
print("\n===== 任务3 =====")
text3 = "ct cat caat caaat"
print("原文：", text3)
print(r"ca*t  :", re.findall(r"ca*t", text3))
print(r"ca+t  :", re.findall(r"ca+t", text3))
print(r"ca?t  :", re.findall(r"ca?t", text3))
print(r"ca{2}t:", re.findall(r"ca{2}t", text3))
print(r"ca{1,2}t:", re.findall(r"ca{1,2}t", text3))
```

# 实验1：正则表达式的基础匹配
## 任务1
文本：`cat on mat; later: cats and dogs`
- `\d+`：匹配一个或多个数字，文本里没有数字，返回空列表
- `\s+`：匹配一段连续空白（空格、制表符等）
- `\bcat\b`：`\b` 是单词边界，只匹配独立单词 `cat`；`cats` 不会被匹配

## 任务2
文本：`cat dog bird CAT`
- `cat|dog`：`|` 代表或，优先匹配左边的 `cat`、`dog`；区分大小写，`CAT` 不匹配
- `re.findall(r"a|aa|aaa", "aaaa")`：按顺序匹配，优先匹配左边短的 `a`，逐个取单字符a
- `re.findall(r"aaa|aa|a", "aaaa")`：优先匹配左边更长的 `aaa`，正则**或分支是顺序优先**，不是最长优先！

## 任务3
文本：`ct cat caat caaat`
- `ca*t`：`*` → a出现0次或多次，可以匹配 `ct`（0个a）、`cat`、`caat`、`caaat`
- `ca+t`：`+` → a出现至少1次，不能匹配 `ct`
- `ca?t`：`?` → a出现0次或1次，匹配 `ct`、`cat`
- `ca{2}t`：`{2}` → a严格等于2个，只匹配 `caat`
- `ca{1,2}t`：`{1,2}` → a出现1~2次，匹配 `cat`、`caat`


```python
import re

# ========== 任务1：捕获组提取邮箱用户名、域名 ==========
text1 = "联系：alice@example.com ; 备选：bob@test.org"
print("===== 任务1 提取邮箱 =====")
# 分组：(\w+) 用户名，@([\w.]+) 域名
pattern_email = r"(\w+)@([\w.]+)"
res1 = re.findall(pattern_email, text1)
print("(用户名, 域名)：", res1)

# ========== 任务2：后向引用找连续重复单词 ==========
text2 = "cat cat, dog cat, bird bird"
print("\n===== 任务2 连续重复单词 =====")
# \b 单词边界，(\w+)捕获单词，\1 回溯引用第1组
pattern_repeat = r"\b(\w+)\b \b\1\b"
res2 = re.findall(pattern_repeat, text2)
print("重复的单词：", res2)

# ========== 任务3：日期格式转换 年-月-日 → 月-日-年 ==========
text3 = "2024-05-01\n2024/05/02\n2024.05.03"
print("\n===== 任务3 日期转换 =====")
print("转换前：")
print(text3)
# 匹配 年 分隔符 月 分隔符 日，分隔符 - / .
pattern_date = r"(\d{4})([-\./])(\d{2})\2(\d{2})"
# 替换为 $3-$4-$1  月-日-年
text3_new = re.sub(pattern_date, r"\3-\4-\1", text3)
print("\n转换后：")
print(text3_new)

# ========== 任务4：电话号码统一格式 (123) 456-7890 ==========
text4 = "联系我：123-456-7890 或 987.654.3210"
print("\n===== 任务4 号码格式化 =====")
print("转换前：", text4)
# 捕获三组数字，分隔符可以是 - 或者 .
pattern_tel = r"(\d{3})[-.](\d{3})[-.](\d{4})"
text4_new = re.sub(pattern_tel, r"(\1) \2-\3", text4)
print("转换后：", text4_new)
```

# 实验2：分组、回溯引用与替换 实验总结
## 任务1：邮箱捕获组提取
正则：`r"(\w+)@([\w.]+)"`
- `()` 代表捕获分组，第1组 `(\w+)` 匹配用户名，第2组 `([\w.]+)` 匹配域名
- `findall` 在存在多个捕获组时，返回元组列表 `(用户名,域名)`
- 结果：`[('alice', 'example.com'), ('bob', 'test.org')]`

## 任务2：回溯引用匹配连续重复单词
正则：`r"\b(\w+)\b \b\1\b"`
- `(\w+)`：捕获一个单词存入分组1
- `\1`：**回溯引用（后向引用）**，代表和第1分组内容完全一样的文本
- 只匹配紧挨着重复两次的单词；`dog cat` 不满足，结果 `['cat', 'bird']`

## 任务3：日期格式转换
正则：`r"(\d{4})([-\./])(\d{2})\2(\d{2})"`
- 分组1：年，分组2：分隔符（- / .），分组3：月，分组4：日
- `\2` 回溯引用分隔符，保证前后分隔符一致
- `re.sub` 替换表达式 `\3-\4-\1`：月-日-年
- 支持三种原始分隔符 `- / .`，换行保留，不会打乱行数

## 任务4：电话号码标准化
正则：`r"(\d{3})[-.](\d{3})[-.](\d{4})"`
- 捕获区号、中间段、尾号三组数字
- 替换模板：`(\1) \2-\3`，统一输出 `(xxx) xxx-xxxx`
- 自动兼容 `-` 和 `.` 两种分隔写法

### 核心知识点
1. `(pattern)`：捕获分组，用于提取内容、后续引用
2. `\数字`：回溯引用，匹配和前面某个分组完全相同的文本
3. `re.sub(正则, 替换模板, 文本)`：支持 `\1 \2` 在替换串里引用分组


```python
from bs4 import BeautifulSoup

html_doc = """
<html><head><title>实验课程</title></head><body>
<h1>数据分析</h1>
<p class="course">第六讲：正则表达式</p>
<p class="course">第七讲：结构化数据</p>
<a href="https://example.org/lecture06">讲义六</a>
<a href="https://example.org/lecture07">讲义七</a>
</body></html>
"""

# ========== 任务1：构建解析树，提取标题、h1 ==========
print("===== 任务1 =====")
soup = BeautifulSoup(html_doc, "html.parser")
title_text = soup.title.get_text()
h1_tag = soup.h1
print("页面标题文本：", title_text)
print("h1标签名：", h1_tag.name)
print("h1文本：", h1_tag.get_text())

# ========== 任务2：find_all 查找class=course的p标签、所有a链接 ==========
print("\n===== 任务2 =====")
p_list = soup.find_all("p", class_="course")
print("class='course' 的段落：")
for p in p_list:
    print("  ", p.get_text())

a_list = soup.find_all("a")
print("所有链接：")
for a in a_list:
    link_text = a.get_text()
    href = a["href"]
    print(f"  链接文字：{link_text}, href={href}")

# ========== 任务3：get_text() 对比不同范围文本 ==========
print("\n===== 任务3 =====")
print("整个文档get_text：")
print(soup.get_text())

print("\n单个段落（第一个course段落）get_text：")
print(p_list[0].get_text())

print("\n单个链接（第一个a标签）get_text：")
print(a_list[0].get_text())
```
